# Jupyter notebook basics

Explore in cells, then restart the kernel and run every cell from the top before sharing. This notebook uses only a small in-memory CSV sample.

Install `numpy`, `pandas`, and `matplotlib` in the active notebook environment if they are not already available.

```python
import gc
from io import StringIO
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

## Compare Python and NumPy

The two cells below perform the same kind of work at the same size. `%timeit` repeats a short expression and reports an average.

```python
%timeit [i * i for i in range(100_000)]
```

```python
%timeit np.arange(100_000) ** 2
```

## Load a CSV, inspect it, and plot a result

This sample is embedded in the notebook so it does not depend on a separate data file. The last expression in a cell is rendered automatically.

```python
csv_text = StringIO('model,accuracy,training_time_seconds\nLinear,0.72,0.1\nRandom Forest,0.89,2.3\nNeural Net,0.94,45.6')
df = pd.read_csv(csv_text)
df
```

```python
ax = df.plot(x='model', y='accuracy', kind='bar', legend=False, title='Model accuracy')
ax.set_ylabel('accuracy')
plt.tight_layout()
plt.show()
```

## Time a whole cell once

`%%time` goes at the top of a cell and reports one elapsed wall-time measurement. Use it for a longer operation such as a training run.

```python
%%time
values = np.random.default_rng(42).random(100_000)
values.mean()
```

## Check and release memory

Notebook variables stay alive between cells. Delete large arrays when you no longer need them, then ask Python's garbage collector to reclaim unreachable objects. Restarting the kernel clears all notebook state.

```python
large = np.zeros(1_000_000, dtype=np.float64)
print(f'Array payload: {large.nbytes / 1_000_000:.1f} MB')
del large
gc.collect()
```

## Share safely

Before sharing, choose **Kernel → Restart & Run All**. Every cell should complete in order without relying on a variable created by an earlier manual run. Keep credentials out of notebook cells and outputs.
