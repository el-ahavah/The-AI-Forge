## Load a CSV and plot it

```python
from io import StringIO
import pandas as pd

csv_text = StringIO("""model,accuracy
Linear,0.72
Random Forest,0.89
Neural Net,0.94
""")

df = pd.read_csv(csv_text)
df
```

```python
import matplotlib.pyplot as plt

ax = df.plot(x="model", y="accuracy", kind="bar", legend=False, title="Model accuracy")
ax.set_ylabel("accuracy")
plt.tight_layout()
plt.show()
```
